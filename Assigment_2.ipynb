# -Assigment 2: Vehicle Gas Mileage Data--
-- Examining vehicle gas mileage data using pandas. The analysis includes inspecting the dataset, cleaning the unnamed vehicle-name column, creating a subset, renaming a column, calculating summary statistics, and filtering for one vehicle type.--

```python
import pandas as pd

# Load the vehicle gas mileage dataset
url = "https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/mtcars.csv"
cars = pd.read_csv(url)

# Display the first five rows
cars.head()

# Display the number of rows and columns
print("Dataset shape:", cars.shape)

# Display the column names
print("\nColumn names:")
print(cars.columns)

# Display information about the dataset
print("\nDataset information:")
cars.info()

# Display descriptive statistics for the numerical columns
cars.describe()

# Rename the unnamed first column as vehicle
cars.rename(columns={cars.columns[0]: "vehicle"}, inplace=True)

# Confirm that the column was renamed
cars.head()

# Create a subset with only the required columns first
cars_subset = cars[["mpg", "hp", "wt", "cyl"]].copy()
cars_subset.head()

# Rename cyl to cylinders on the newly created subset
cars_subset.rename(columns={"cyl": "cylinders"}, inplace=True)

# Display the updated subset
cars_subset.head()


```

**Summary Stats: count, mean, SD, min, max, and quartile values for gas mileaga, horsepower, vehicle weight, and number of cylinders.**

```python
# Calculate summary statistics for the subset
cars_subset.describe()

# Calculate the average miles per gallon
average_mpg = cars_subset["mpg"].mean()
print(f"Average MPG for all vehicles: {average_mpg:.2f}")
```

**4-Cylinder Vehicles Summary Stats**


```python
# Filter the data to include only four-cylinder vehicles
four_cylinder_cars = cars_subset[cars_subset["cylinders"] == 4]

# Display the filtered data
four_cylinder_cars

# Calculate summary statistics for four-cylinder vehicles
four_cylinder_cars.describe()

# Calculate the average MPG for four-cylinder vehicles
four_cylinder_average_mpg = four_cylinder_cars["mpg"].mean()
print(f"Number of four-cylinder vehicles: {len(four_cylinder_cars)}")
print(f"Average MPG for four-cylinder vehicles: {four_cylinder_average_mpg:.2f}")
```

**Conclusion
The complete dataset has an average gas mileage of approximately 20.09 MPG. After filtering the dataset, the 11 four-cylinder vehicles have an average gas mileage of approximately 26.66 MPG. Therefore, the four-cylinder vehicles in this dataset have a higher average MPG than the overall group of vehicles.**
