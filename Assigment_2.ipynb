# Assignment 2: Vehicle Gas Mileage Data

Examining vehicle gas mileage data using pandas. The analysis includes inspecting the dataset, cleaning the unnamed vehicle-name column, creating a subset, renaming a column, calculating summary statistics, and filtering for one vehicle type.

```python
import pandas as pd

# Load the vehicle gas mileage dataset
url = "https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/mtcars.csv"
cars = pd.read_csv(url)

# Display the first five rows
display(cars.head())

# Display the number of rows and columns
print("Dataset shape:", cars.shape)

# Display the column names
print("\nColumn names:", cars.columns.tolist())

# Display information about the dataset
print("\nDataset information:")
cars.info()

# Display descriptive statistics for the numerical columns
display(cars.describe())

# Rename the unnamed first column as vehicle
cars.rename(columns={cars.columns[0]: "vehicle"}, inplace=True)

# Create a subset with only the required columns
cars_subset = cars[["mpg", "hp", "wt", "cyl"]].copy()

# Rename cyl to cylinders on the subset
cars_subset.rename(columns={"cyl": "cylinders"}, inplace=True)

# Display the updated subset first five rows
display(cars_subset.head())
```

### Summary Statistics for the Complete Subset (mpg, hp, wt, cylinders)

```python
# Calculate and display summary statistics for the subset
display(cars_subset.describe())
```

### Filtered Subset: 4-Cylinder Vehicles Summary Statistics

```python
# Filter the data to include only four-cylinder vehicles
four_cylinder_cars = cars_subset[cars_subset["cylinders"] == 4]

# Display summary statistics for four-cylinder vehicles
display(four_cylinder_cars.describe())
```

### Data-Driven Insights & Conclusions

Based on the summary statistics of the dataset, we can draw the following key conclusions:
- **Fuel Efficiency vs. Engine Size**: Four-cylinder vehicles are significantly more fuel-efficient than the overall fleet, average-performing at approximately **26.66 MPG** compared to the general average of **20.09 MPG**.
- **Weight and Horsepower Trade-offs**: The increased efficiency of the four-cylinder group is directly coupled with lower weight (averaging **2.29 thousand lbs** vs. **3.22 thousand lbs** overall) and lower horsepower (averaging **82.64 HP** vs. **146.69 HP** overall).
